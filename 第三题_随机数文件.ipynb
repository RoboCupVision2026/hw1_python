# 第三题：随机数文件

**题目：** 创建文件data.txt，文件共100000行，每行存放一个随机的1～100之间的整数。

```python
from random import randint
from pathlib import Path
fp=Path("data.txt")
with fp.open("w",encoding="utf-8")as f:
    for _ in range(100000):
        n=randint(1,100)
        f.write(str(n)+"\n")

```

先从random里面取出randint这个函数，用来随机生成整数，用法是randint(起始，结束)，是闭区间。然后再从pathlib里面取出Path，**注意这里的Path第一个字母要大写**。然后再调用Path对文件data **（这里的data文件要用双引号括起来）** 创建一个记录文件路径的变量fp，这里的文件路径默认是和这个代码的运行环境有关，这里不展开叙述。然后再使用with对这个文件进行操作，open打开文件，"w"表示对这个文件进行写入，不管这个文件有没有被创建，里面有什么，在这里了的时候都被创建了，并且清空了。然后encoding="utf-8"保证文件内容用转化成字节表示的时候采用的编码，也可省略。我们这里用f临时表示这个文件。后面有一个循环，先生成一个随机整数，然后再用write函数写入这个文件里面，注意这个write函数里面只能是一个字符串，所以应该先转化成字符串然后再用+把两个连接起来。
