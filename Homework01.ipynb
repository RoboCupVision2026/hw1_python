```python
#对于删去列表中的奇数元素，有多种方法，下面给出三种方法：
#其中法一比较另类，是我自己想的，
#法二法三借助编译器自带的ai工具完成
#法一：
my_list = list(range(1000))
for idx in range(len(my_list)):
    if my_list[idx] % 2 == 1:
        my_list[idx] = 'a'
while 'a' in my_list:
    my_list.remove('a')
print(my_list)
#法二：逆序删除
my_list2 = list(range(1000))
for idx in range(len(my_list2)-1,-1,-1):
    if my_list2[idx] % 2 == 1:
        del my_list2[idx]
print(my_list2)
#法三：列表推导式
my_list3 = list(range(1000))
my_list3 = [i for i in my_list3 if i % 2 == 0]
print(my_list3)
```

```python
#创建顺序文件
import random
import os
import string

chars = string.ascii_letters + string.digits
os.makedirs("img", exist_ok=True)

file = set ()
while  len(file) < 100:
    file_name = ''.join(random.choices(chars,k = 4))
    if file_name not in file:
        file.add(file_name)
        file_path = os.path.join("img",f"{file_name}.png")
        open(file_path,"w").close()
```

```python
import os
import random

base_path = 'img'
all_files = [f for f in os.listdir('img') if f.endswith('.png') ]
#列表推导式
#等价于下面的for循环:
#all_files = []
#for f in os.listdir(base_path):
#（os.listdir以列表形式返回文件中所有文件名）
#    if f.endswith('.png'):
#    all_files.append(f)
rename_files = random.sample(all_files,50)
for file_name in rename_files:
    old_path = os.path.join(base_path,file_name)
    new_path = os.path.join(base_path,file_name[:-4]+'.jpg')
    os.rename(old_path,new_path)
#os.path.join是拼接路径函数，加斜杠
#file_name此时为字符串[:-4]，删去最后四个字符
#最后重命名
```

```python
import random

number02 = list()
with open("十行三列随机数.txt",'w') as f:
    i = 0
    for i in range(10):
        f.write(str(random.randint(1,100)) + ',')
        a = random.randint(1,100)
        f.write(str(a) + ',')
        f.write(str(random.randint(1,100)) + '\n')
        number02.append(a)
average = sum(number02)/10
print("第二列最大值为:",max(number02))
print("第二列最小值为:",min(number02))
print("第二列平均值为:",average)
number02.sort()
b = (number02[4]+number02[5])/2
print("第二列中位数为:",b)
```

```python
class Person:
    def __init__(self,name,age,gender):
        self.name = name
        self.age =age
        self.gender = gender
    
    def personinfo(self):
        print(f"姓名：{self.name}")
        print(f"年龄：{self.age}")
        print(f"性别：{self.gender}")


class Student(Person):
    def __init__(self,name,age,gender,college_name,class_name):
        super().__init__(name,age,gender)
        self.college_name = college_name
        self.class_name = class_name

    def personinfo(self):
        print(f"姓名：{self.name}")
        print(f"年龄：{self.age}")
        print(f"性别：{self.gender}")
        print(f"学院：{self.college_name}")
        print(f"班级：{self.class_name}")
    def __str__(self):
        return(f"姓名：{self.name}"
               f"年龄：{self.age}"
               f"性别：{self.gender}"
               f"学院：{self.college_name}"
               f"班级：{self.class_name}")
s1 = Student("梅冬瑞\n","18\n","男\n","机械\n","具身智能2601\n")
print(s1)
```
