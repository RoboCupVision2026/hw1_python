# 第二题：文件复制

**题目：** 在工作目录下写程序创建一个文件test.txt，并随机生成用户指定行数的内容，包含ASCII标准的任意非控制字符。写程序实现该文件的拷贝，拷贝文件命名为copy_test.txt。

```python
from pathlib import Path
from random import randint
current = Path.cwd()
file_path = current.resolve()/"test.txt"
hang = int(input("请输入行数："))
with open(file_path,"w",encoding = "utf-8")as file:
    for i in range(hang):
        input_1 =""
        for j in range(80):
            number = randint(32,126)
            input_1 = input_1 + chr(number)
        file.write(input_1+"\n")
new = file_path.resolve().parent/"copy_test.txt"
with open(new,"w",encoding = "utf-8")as file1:
    with open(file_path,"r",encoding = "utf-8")as file2:
        for line in file2:
            file1.write(line)





```
