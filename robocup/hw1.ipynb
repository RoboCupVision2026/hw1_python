第六题：找bug
原因：python列表长度可变，删除一个元素之后长度变小，len(list)长度开始就确定了为1000，动态列表项数前移（如list[1]变成list[0]）使遍历次数超出列表元素个数，发生报错。修改后代码：

```python
if __name__ == "__main__":
    my_list = list(range(1000))
    #倒序输出避免下标变化
    for idx in range(len(my_list)-1,-1,-1):
        if my_list[idx]%2==1:
            my_list.pop(idx) 
```

第七题：最后剩下谁

```python
people=list(range(1,234))
idx=0
while len(people)>1:
    #%是为了当（idx+2）>len时不超出列表元素
    idx=(idx+2)%len(people)
    people.pop(idx)

print(f"最后剩下{people[0]}号")


```

第九题：创建顺序文件

```python

import os
import random
import string
from pathlib import Path

# 获取当前脚本所在目录，并拼接出 img 目录的路径
base_dir = Path(__file__).parent
img_dir = base_dir / 'img'

# 如果 img 目录不存在，则创建它
img_dir.mkdir(exist_ok=True)

# 用于存放已生成的文件名，防止重复
existing_names = set()

while len(existing_names) < 100:
    # 随机生成 4 位由大写字母和数字组成的字符串
    chars = string.ascii_uppercase + string.digits
    filename = ''.join(random.choices(chars, k=4)) + '.png'
    
    # 如果这个文件名还没生成过，则加入集合并创建文件
    if filename not in existing_names:
        existing_names.add(filename)
        file_path = img_dir / filename
        # touch() 方法用于创建一个空文件
        file_path.touch()

print(f"已在 {img_dir} 目录下成功创建 100 个 .png 文件。")
```

第十题：文件改名

```python
import random
from pathlib import Path

# 获取 img 目录路径
img_dir = Path(__file__).parent / 'img'

# 获取 img 目录下所有以 .png 结尾的文件列表
png_files = list(img_dir.glob('*.png'))

# 随机抽取 50 个文件
files_to_rename = random.sample(png_files, 50)

# 遍历这 50 个文件并进行重命名
for file in files_to_rename:
    # with_suffix 方法会自动替换文件后缀名
    new_path = file.with_suffix('.jpg')
    # 重命名文件
    file.rename(new_path)

print("已成功将 50 个文件的后缀名修改为 .jpg。")
```

第11题：文件输入

```python
import random
import statistics

# 1. 生成并写入文件
with open('data_10x3.txt', 'w', encoding='utf-8') as f:
    for _ in range(10):
        # 生成3个 1~100 的随机浮点数（保留两位小数）
        row = [f"{random.uniform(1, 100):.2f}" for _ in range(3)]
        f.write(','.join(row) + '\n')

# 2. 读取并提取第二列数据
col2_data = []
with open('data_10x3.txt', 'r', encoding='utf-8') as f:
    for line in f:
        parts = line.strip().split(',')
        if len(parts) >= 2:
            col2_data.append(float(parts[1]))

# 3. 计算并输出结果
print("第二列数据:", col2_data)
print(f"最大值: {max(col2_data)}")
print(f"最小值: {min(col2_data)}")
print(f"平均值: {sum(col2_data) / len(col2_data):.2f}")
print(f"中位数: {statistics.median(col2_data)}")
```

第12题：文件复制

```python
import random
import string

n = int(input("请输入要生成的行数: "))

# 1. 创建 test.txt 并写入随机字符
with open('test.txt', 'w', encoding='utf-8') as f:
    for _ in range(n):
        # ASCII非控制字符：可见字符（包括字母、数字、标点、空格），范围大致在 32~126
        line_length = random.randint(10, 40) # 每行长度随机
        line_content = ''.join(chr(random.randint(32, 126)) for _ in range(line_length))
        f.write(line_content + '\n')

print("test.txt 创建完成。")

# 2. 拷贝文件
with open('test.txt', 'r', encoding='utf-8') as f_src, \
     open('copy_test.txt', 'w', encoding='utf-8') as f_dst:
    for line in f_src:
        f_dst.write(line)

print("拷贝完成，文件名为 copy_test.txt。")
```

第15题：文件的批量创建

```python
import os
import random
import string

i = int(input("请输入要创建的文件数量: "))
j = int(input("请输入每个文件的行数: "))

dir_name = 'test'

# 1. 创建目录
os.makedirs(dir_name, exist_ok=True)

# 2. 批量创建文件并写入内容
for k in range(1, i + 1):
    file_path = os.path.join(dir_name, f"file_{k}.txt")
    with open(file_path, 'w', encoding='utf-8') as f:
        for _ in range(j):
            line = ''.join(random.choices(string.ascii_letters, k=random.randint(10, 20)))
            f.write(line + '\n')

print(f"已在 {dir_name} 目录下创建 {i} 个文件。")

# 3. 遍历目录，修改文件名和文件内容
for filename in os.listdir(dir_name):
    old_file_path = os.path.join(dir_name, filename)
    
    # 修改文件内容：在每一行后加 -python
    with open(old_file_path, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    
    with open(old_file_path, 'w', encoding='utf-8') as f:
        for line in lines:
            # rstrip('\n') 去掉原有的换行，加上 -python 后再补上换行
            f.write(line.rstrip('\n') + '-python\n')
            
    # 修改文件名：利用 os.path.splitext 分离文件名和后缀
    name, ext = os.path.splitext(filename)
    new_filename = f"{name}-python{ext}"
    new_file_path = os.path.join(dir_name, new_filename)
    
    os.rename(old_file_path, new_file_path)

print("已完成文件名和内容的修改，所有名称和行尾已追加 '-python'。")
```

第16题：类的继承与魔术方法

```python
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender=gender
    
    #输出人的信息
    def personInfo(self):
        print(f"姓名：{self.name}，年龄：{self.age},性别:{self.gender}")

class Student(Person):
    def __init__(self,name,age,gender,college,clas):
        super().__init__(name,age,gender)
        self.college=college
        self.clas=clas

    #继承父类方法
    def personInfo(self):
            super().personInfo()
            print(f"学院：{self.college}，班级{self.clas}")

    #重写__str__法，返回学生信息
    def __str__(self):
         return f"姓名：{self.name}，年龄：{self.age},性别:{self.gender},学院：{self.college}，班级{self.clas}"


# ================= 测试代码 =================
if __name__ == '__main__':
    print("--- 测试 Person 类 ---")
    p = Person("张三", 25, "男")
    p.personInfo()

    print("\n--- 测试 Student 类 ---")
    s = Student("李四", 20, "女", "计算机学院", "软件工程2班")
    
    # 调用重写后的 personInfo 方法
    s.personInfo()
    
    # 测试 __str__ 方法（直接打印对象时触发）
    print("\n--- 测试 __str__ 魔术方法 ---")
    print(s)
         
```
