```python
people = list(range(1, 234))
index = 0
while len(people)>1:
    index = (index + 2) % len(people)
    people.pop(index)
print(people[0])

```

```python
import random

with open("doc_write.txt", "w", encoding="utf-8") as f:
    for i in range(10):
        nums = [random.randint(0, 100) for _ in range(3)]
        line = ",".join(str(num) for num in nums)
        f.write(line + "\n")




with open("doc_write.txt", "r", encoding="utf-8") as f:
   
    col2_list = []
    

    for line in f:
        # strip() 去掉每行末尾的换行符，split(",") 按逗号把一行切割成列表
        parts = line.strip().split(",")
        num = int(parts[1])
        col2_list.append(num)

#
max_val = max(col2_list)               
min_val = min(col2_list)               
avg_val = sum(col2_list) / len(col2_list)  


sorted_list = sorted(col2_list)
n = len(sorted_list)

median = (sorted_list[n//2 - 1] + sorted_list[n//2]) / 2


print(f"第二列所有数字：{col2_list}")
print(f"最大值：{max_val}")
print(f"最小值：{min_val}")
print(f"平均值：{avg_val:.2f}")  # .2f 表示保留两位小数
print(f"中位数：{median}")

```

```python
class Person:
    def __init__(self,name,age,gender):
        self.name = name
        self.gender = gender
        self.age =age
    def personInfo(self):
        print(f"姓名是{self.name}")
        print(f"年龄是{self.age}")
        print(f"性别是{self.gender}")

class Student(Person):
    def __init__(self,name,age,gender,college,class_name):
        super().__init__(name,age,gender)
        self.college=college
        self.class_name=class_name

    def studentInfo(self):
        super().personInfo()
        print(f"学院是{self.college}")
        print(f"班级是{self.class_name}")

    def __str__(self):
        return (f"姓名是{self.name}\n年龄是{self.age}\n性别是{self.gender}\n学院是{self.college}\n班级是{self.class_name}")

student=Student("张三",20,"男","计算机学院","计算机科学与技术")
student.studentInfo()
print("")#print 自带换行
print(student)
```

```python
def fibonacci(n):
    if n == 1:
        return 0
    if n == 2:
        return 1
    if n == 3:
        return 1
    return fibonacci(n-1) + fibonacci(n-2)
fibonacci(20)
```

```python
from pathlib import Path
import random
import string

base_dir = Path.cwd()

img_dir = base_dir / "img"
img_dir.mkdir(exist_ok=True)

png_files = list(img_dir.glob("*.png"))
used_names = {p.name for p in img_dir.iterdir()}

while len(png_files) < 100:
    
    name = "".join(random.choices(string.ascii_uppercase + string.digits, k=4)) + ".png"

   
    if name in used_names:
        continue

    
    file_path = img_dir / name
    file_path.touch()

    png_files.append(file_path)
    used_names.add(name)

```
