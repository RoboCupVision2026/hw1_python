```python
import os
import random
lis=[]
for i in range(32,127):
    lis.append(chr(i))
dir_root='test'
os.makedirs('test',exist_ok=True)
n=int(input('请输入文件数'))
j=int(input("请输入行数"))
for i in range(n):
    with open(os.path.join(dir_root,'file_'+str(i+1)),'w',encoding='utf-8')as f:
        for i in range(j):
            chars=''.join(random.choices(lis,k=10))
            f.write(f"{chars}\n")


lis_dir=os.listdir('test')
for old in lis_dir:
    new=old+'python'
    old_path=os.path.join('test',old)
    new_path=os.path.join('test',new)
    os.rename(old_path,new_path)
    ls=[]
    with open(new_path,'r',encoding='utf-8') as f:
        for line in f:
            ls.append(line.strip())
    with open(new_path,'w',encoding='utf-8')as f:
        for i in ls:
            f.write(f'{i}python\n')
```
