## 第六题：找bug
### 问题在于他把索引从前向后添加，导致删除数字后，数组自动前移，补齐位置，不仅少检查了数，到最后检查到空索引还会报错。然后命名时list已经被占用


```python
numbers= list(range(1000))
for idx in range(len(numbers) - 1, -1, -1):
    if numbers[idx] % 2 == 1:
        numbers.pop(idx)
print(numbers)            


```

## 或者让他从倒着执行删除  如下

"然后我发现其实并不行qwq。发给ai他说是因为列表长度变了，索引还是会读串。他又加了一个变量，感觉这样也行\^^/"

```python
numbers= list(range(1000))
length= len(numbers)
for idx in range(length):
    position= length - 1 - idx
    if numbers[position] % 2 == 1:
        numbers.pop(position)
```

## 第七题



```python
people= list(range(1, 234))
idx= 0
while len(people) > 1:
    idx= idx + 2
    idx= idx % len(people)  #"依次向前删除，加上除余考虑最后一位循环报数到第一位的情况"
    people.pop(idx)
print("最后剩下的是原来的", people[0], "号")
```

##  十一

```python
import random
with open("date.txt", "w", encoding="utf-8") as file:
    for _ in range(10):
        a= random.randint(1, 100)
        b= random.randint(1, 100)
        c= random.randint(1, 100)
        row= f"{a},{b},{c}\n"
        file.write(row)
print("date.txt 已生成")
nums= []
with open("date.txt", "r", encoding="utf-8") as file:
    for line in file:
        columns= line.split(",")
        value= int(columns[1])
        nums.append(value)
print("最大值", max(nums))
print("最小值",min(nums))
import statistics
print("平均数",  statistics.mean(nums))
print("中位数",  statistics.median(nums))
```

# 最后一题

```python
class Person:
    def __init__(x, name, age, sex):
        x.name= name
        x.age= age
        x.sex= sex
    def personInfo(x):
        print("名字", x.name)
        print("年龄", x.age)
        print("性别", x.sex)
class Student(Person):
    def __init__(x, name, age, sex, college, classroom):
        super().__init__(name, age, sex)
        x.college= college
        x.classroom= classroom
    def personInfo(x):
        super().personInfo()
        print("学院", x.college)
        print("班级", x.classroom)
    def __str__(x):
        return f"名字{x.name}，年龄{x.age}，性别{x.sex}，学院{x.college}，班级{x.classroom}"
name= input("姓名：")
age= input("几岁：")
sex= input("男/女：")
college= input("学院：")
classroom= input("班级：")
Student(name, age, sex, college, classroom).personInfo()
print(Student(name, age, sex, college, classroom))
```
