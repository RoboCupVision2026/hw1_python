# 第三题：类的继承与魔术方法

**题目：** 1. 创建Person类，属性有姓名、年龄、性别，创建方法personInfo，打印这个人的信息。
2. 创建Student类，继承Person类，属性有学院college、班级class，重写父类personInfo方法，除了调用父类方法打印个人信息外，将学生的学院、班级信息也打印出来。
3. 重写__str__方法，返回student的信息。

```python
class Person :
    def __init__(self,name,age,sex):
        self.name = name 
        self.age = age
        self.sex = sex
    def personInfo(self):
        print(f"姓名：{self.name}")
        print(f"年龄：{self.age}")
        print(f"性别：{self.sex}")
class Student(Person):
    def __init__(self, name, age, sex,college,classs):
        super().__init__(name, age, sex)
        self.college = college
        self.classs = classs
    def personInfo(self):
        super().personInfo()
        print(f"学院：{self.college}")
        print(f"班级：{self.classs}")
    def __str__(self):
        return(
            f"姓名：{self.name}\n"
            f"年龄：{self.age}\n"
            f"性别：{self.sex}\n"
            f"学院：{self.college}\n"
            f"班级：{self.classs}"
        )




```
