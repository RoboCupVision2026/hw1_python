# 第六题：找 bug
**题目：** 下面这些代码的目的是把 list 中的所有奇数删除，但是这段代码却无法运行，请你找出来并说明 bug 在哪里、为什么发生，并进行改正。

```python
if _name_ == "_main_":
    list = list(range(1000))
    for idx in range(len(list)):
        if list[idx] % 2 == 1:
            list.pop(idx)
```

## 三个 bug 及原因

**Bug 1：`_name_` 写错了**
应是 `__name__`（前后各**两个**下划线），`"_main_"` 也应是 `"__main__"`。

**Bug 2：用 `list` 当变量名，覆盖了内置函数**
`list` 是 Python 内置函数（用来构造列表），把它当变量名后，`list` 就不再指向内置函数了。应换个名字，比如 `numbers`。

**Bug 3（核心）：边遍历边 `pop` 删除元素**
`range(len(list))` 在循环**开始时就固定成 1000 次**，但 `pop` 会让列表**越来越短**。结果有两个坏处：
- **下标越界**：`idx` 一直涨到 999，列表却缩到 500，`list[idx]` 就报 `IndexError`（实测崩溃在 idx=501 时，长度只剩 500）。
- **跳过元素**：删掉一个元素后，后面的元素会左移一位，导致某些元素被跳过没检查。

````python
number = list (range(1000))
for i in list (range(1000))[::-1]:
    if i % 2 != 0:
        number.remove(i)
````
