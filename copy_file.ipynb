```python
import random
lis=[]
n=int(input('请输入行数'))
for i in range(32,127):
    lis.append(chr(i))
with open('test.txt','w')as f:
    for i in range(n):
        f.write(f'{random.choice(lis)}\n')
with open('test.txt','r',encoding='utf-8')as f1 ,open('copy_test.txt','w',encoding='utf-8')as f2:
    for lines in f1:
        f2.write(lines)
```
