```python
"Bug:在遍历列表时使用 pop() 修改了列表长度,range(len(list)) 一开始固定为 0-999,但删除元素后列表会缩短、后面的元素会向前移动，最终导致索引越界，也可能漏检元素。"
if __name__ == "__main__":
    my_list = list(range(1000))
    for idx in range(len(my_list)-1, -1, -1):
        if my_list[idx] % 2:
            my_list.pop(idx)
    print(my_list)
```

```python
#第一题 文件输入
import random
import statistics
rows = [[random.randint(1,100) for i in range(3)] for j in range(10)]
with open("data.txt","w",encoding="utf-8") as f:
    for row in rows:
        f.write(",".join(map(str,row)) + "\n")
with open("data.txt","r",encoding="utf-8") as f:
    second_column =[int(line.strip().split(',')[1]) for line in f]
print("第二列的最大值为：",max(second_column))
print("第二列的最小值为：",min(second_column))
print("第二列的平均值为：",statistics.mean(second_column))
print("第二列的中位数为：",statistics.median(second_column))
```

```python
#第二题 文件复制
import shutil
import random
import string
from pathlib import Path
i = int(input("请输入要生成的行数："))
with open("test.txt","w",encoding="utf-8") as f:
    for _ in range(i):
        random_line =''.join(random.choices(string.printable[:95], k=3))
        f.write(random_line+'\n')
shutil.copy("test.txt", "copy_test.txt")
#第三题 文件修改
with open("test.txt","r",encoding="utf-8") as f:
    content = f.read()
new_content =  "python" + content + "python"
with open("test.txt","w",encoding="utf-8") as f:
    f.write(new_content)
#第四题 文件对比
with open("test.txt","r",encoding="utf-8") as f1, \
     open("copy_test.txt","r",encoding="utf-8") as f2:
    for i, (line1, line2) in enumerate(zip(f1, f2), start=1):
        if line1 != line2:
            print(f"第{i}行不同\n")
```

```python
#第五题 文件批量创建
import random
import string
from pathlib import Path
i=int(input("请输入要生成的文件数："))
j=int(input("请输入要生成的随机字符行数："))
test_dir = Path("test")
test_dir.mkdir(exist_ok=True)
for k in range(i):
    file_path = test_dir / f"file_{k}.txt"
    content = ""
    for _ in range(j):
        line = "".join(random.choices(string.ascii_letters + string.digits, k=random.randint(5,15)))
        content += line + '\n'
        file_path.write_text(content,encoding="utf-8")
for file in test_dir.iterdir():
    if not file.is_file():
        continue
    lines = file.read_text(encoding="utf-8").splitlines()
    new_content = "\n".join(line + "-python" for line in lines) + "\n"
    file.write_text(new_content, encoding="utf-8")
    file.rename(test_dir/f"{file.stem}-python.txt")   
```

```python
#类、对象、属性、方法
class Person:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender=gender
    def personInfo(self):
        print(f"姓名：{self.name}")
        print(f"年龄：{self.age}")
        print(f"性别：{self.gender}")
# 继承、方法重写、新增方法
class Student(Person):
    def __init__(self,name,age,gender,college,class_name):
        super().__init__(name,age,gender)
        self.college=college
        self.class_name=class_name
    def personInfo(self):
        super().personInfo()
        print(f"学院：{self.college}")
        print(f"班级：{self.class_name}")
# 把对象变成友好字符串
def __str__(self):
    return (f"姓名：{self.name}\n"
            f"年龄：{self.age}\n"
            f"性别：{self.gender}\n"
            f"学院：{self.college}\n"
            f"班级：{self.class_name}")
# 创建实例化对象: student = Student("张三", 20, "男", "计算机学院", "1班")
# 对象调用类中方法: student.personInfo() 
# 父类已有同名方法，子类重新定义：方法重写
# 父类没有该方法，子类直接定义：新增方法
# 新增的方法只属于子类，不会添加到父类中
```
