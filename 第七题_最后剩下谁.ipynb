# 第七题：最后剩下谁

**题目：** 有233个人围成一圈，顺序排号。从第一个人开始报数（从1到3报数），凡报到3的人退出圈子，问最后留下的是原来第几号的那位。

```python
people = list(range(1,234))
count = 1
j = 0
while len(people) > 1:
    if count == 3:
        count = 0
        people.pop(j)
        j -= 1
    count += 1
    j += 1
    if j == len(people):
        j = 0
print(f"最后剩下了:{people[0]}")
    
```
