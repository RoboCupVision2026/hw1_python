```python
class Person:
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender

    def personInfo(self):
        print(self.name , self.age , self.gender)


class Student(Person):
    def __init__(self, name, age, gender,college,class1):
        super().__init__(name, age, gender)
        self.college = college
        self.class1 = class1

    def personInfo(self):
        super().personInfo()
        print(self.college , self.class1)

    def __str__(self):
        return f"{self.name} , {self.age} , {self.gender} , {self.college} , {self.class1}"


database = [Student("薛佩铉","18","男","能动学院","能动2601")]

name = input("请你输入要查找的学生姓名：")

for stu in database:
    if stu.name == name:
        stu.personInfo()
        break
    else:
        print("很抱歉，查无此人")
```
