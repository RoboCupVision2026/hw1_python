```python
import random
import string

lines = int(input("你想生成的行数："))
num = int(input("你想每行生成多少个字符："))

char = string.ascii_letters + string.digits + string.punctuation

with open('test.txt','w') as f:
    for i in range(lines):
        n = ''.join(random.choices(char,k = num))
        f.write(n + '\n')

with open('test.txt','r') as f1:
    with open('copy_test.txt','w') as f2:
        for line in f1:
            f2.write(line)
```
