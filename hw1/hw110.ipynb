```python
import os
import random

folder = "img"
rename = random.sample(os.listdir(folder),50)

for old in rename:
    
    if not old.endswith(".png"):
        continue

    new = old.replace(".png",".jpg")
    old_path = os.path.join(folder , old)
    new_path = os.path.join(folder , new)
    os.rename(old_path , new_path)
```
