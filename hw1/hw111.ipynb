```python
import random

with open('data1.txt','w') as f:
    for i in range(10):
        num1 = random.randint(1,10)
        num2 = random.randint(1,10)
        num3 = random.randint(1,10)
        f.write(f"{num1},{num2},{num3}\n")

data_2 = []

with open('data1.txt','r') as f:
    for line in f:
        part2 = line.strip().split(',')
        data_2.append(int(part2[1]))

max_2 = max(data_2)
min_2 = min(data_2)
average = sum(data_2)/len(data_2)

data_2.sort()
n = len(data_2)

if n % 2 == 1:
    mid = data_2[n // 2]
else:
    mid_1 = data_2[n // 2 - 1]
    mid_2 = data_2[n // 2]
    mid = (mid_1 + mid_2) / 2

print(max_2 , min_2 , average , mid)
```
