# 16 类的继承与魔术方法

题目：创建Person类，包含姓名、年龄、性别及personInfo方法；Student继承Person，增加学院和班级，重写personInfo并调用父类方法，再重写__str__返回学生信息。

考点：继承、方法重写、super、__str__

```python
class Person:
    def __init__(self,name,age,sex):
        self.name=name
        self.age=age
        self.sex=sex
    def personInfo(self):
        print(self.name,self.age,self.sex)
class Student(Person):
    def __init__(self,name,age,sex,college,cls):
        super().__init__(name,age,sex)
        self.college=college
        self.cls=cls
    def personInfo(self):
        super().personInfo()
        print(self.college,self.cls)
    def __str__(self):
        return f"{self.name} {self.age} {self.sex} {self.college} {self.cls}"
if __name__=="__main__":
    s=Student("张三",20,"男","计算机学院","1班")
    s.personInfo()
    print(s)
```
